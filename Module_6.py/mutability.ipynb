```python
# Mutability or Immutability
# Mutability is the ability or value of value or data to be modified or changed
# Immutability is the reverse of it

# Lists are mutable
# Tuples and Strings are immutable
```

```python
# String
s1 = "Python is fun"
s1 = s1.replace("Python", "Java")
print(s1)
s2 = s2.replace("Python", "Java")
print(s2)
```

```python
# Tuple
t1 = ("Mango", "Orange", "Apple")
t1.append("Banana")
print(t1)
```

```python
# List
l1 = ["Mango", "Orange", "Apple"]
print(id(l1))
l1.append("Banana")
print(l1)
print(id(l1))
```

```python
# List
l1 = ["Mango", "Orange", "Aple"]
print(l1)
l1[-1] = "Apple"
print(l1)
```
