```python
student1 = {"maths":80.5, "eng":76.0, "phy":89.0}
print(student1)
```

```python
# fetch the marks for phy
print(student1["phy"])

# get() : does the same
print(student1.get("phy"))
```

```python
# Difference btw normal fetching and get()
# normal fetching
print(student1["chem"])
```

```python
# get() fetching
print(student1.get("chem"))
```

```python
print(student1.get("chem", 40.0))
```

```python
emp1 = {'id': 1001, 'name':'John', 'salary':10000}
print(emp1.get('phone', 9876543210))

print(emp1.get('id', 9876543210)) #still gets the smae value as it the dict
```

```python
# membership operator => in
print('name' in emp1)
print('phone' in emp1)
print('1001' in emp1) # value part is always false whereas key part is true and only true if it is present in the dict
```

```python
# Updation
sem1_marks = {'maths':78.5, 'eng':71.0, 'phy':86.5}
sem2_marks = {'chem': 81.5, 'bio': 90.5}

sem1_marks.update(sem2_marks)
print(sem1_marks)

# One more example with a little variation
groceries_1 = {'milk': 60, 'rice': 100, 'biscuits': 20}
groceries_2 = {'rice': 110, 'bread': 30}

groceries_1.update(groceries_2) # value of rice gets updated
print(groceries_1)
```

```python
# pop() :deletes that particular key-value pair
groceries_1 = {'milk': 60, 'rice': 100, 'biscuits': 20, 'rice': 110, 'bread': 30}
groceries_1.pop("milk")
print(groceries_1)
```

```python
# duplication of key with diffrent values
# Here, python reads dict from Left-to-Right 
# So, the first key value pair it reads  is considered
groceries_1 = {'milk' : 60, 'rice' : 100, 'biscuits' : 20, 'milk' : 65}
print(groceries_1)
```
