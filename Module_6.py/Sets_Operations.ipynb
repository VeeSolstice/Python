```python
nums = {1, 3, 2, 0, -1}
```

```python
# Membership operator- in, not in
print(0 in nums)
print(10 not in nums)
```

```python
# We cannot do indexing and slicing of sets
```

```python
# we cannot do concatenation of sets
```

```python
# We cannot do repetition like strings,tuples, lists.
```

```python
# Type-Casting  
weekdays = ("Monday", "Tuesday", "Wednesday", "Thursday", "Friday")
weekdays = set(weekdays)
print(weekdays, type(weekdays))
```

```python
# Sets are mutable
```

```python
# To add something 
# add()
set1 = {2, 0, -1}
print(set1)
set1.add(5)
print(set1)

set1.add(2)
print(set1) # No change in the output
```

```python
# To remove something
# remove()
set1 = {2, 0, -1}
print(set1)
set1.remove(0)
print(set1)

# discard()
# The difference between the remove() and the discrad() function 
# is that in case of remove() if u try 
# to remove any element that is not present in the set 
# it gives an error 
# while incase of discard 
# it does not throw any such error
set1.discard(21)
print(set1)
```
