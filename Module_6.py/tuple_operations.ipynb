```python
"""
concatenation, repetition, membership
count, index
min, max, sum
"""
student_detail1 = (1001, "John")
student_detail2 = (78.5, 91.0, 83.5, 79.5)
```

```python
# +
student_details = student_detail1 + student_detail2
print(student_details)
```

```python
# *
t1 = ("Class 5", 5000)
print(t1 * 3)
```

```python
# in, not in
print(91.0 in student_detail2)
print(92.0 not in student_detail2)
```

```python
#  count()
t1 = (10, 1, 4, 9, 1, 0, 3, 1)
# tuple.count(element)
print(t1.count(1))
```

```python
# index()
# tuple.index(element)
print(t1.index(4))  # What is the index of 4 in the tuple 1?
```

```python
# min(tuple)
# max(tuple)
# sum(tuple)
print(f"Smallest number: {min(t1)}") #0
print(f"Biggest number: {max(t1)}")  #10
print(f"Total: {sum(t1)}")           #29
```
