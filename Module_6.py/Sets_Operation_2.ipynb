```python
student1 = {"English", "Maths", "CS", "Chemistry", "Physics"}
student2 = {"English", "Biology", "Chemistry", "Physics"}

print(student1, type(student1))
print(student2, type(student2))
```

```python
# Common subjects of student1 and student2 - intersection
common_subjects = student1.intersection(student2)
print(common_subjects) 

# or 
common_subjects = student1 & student2 
print(common_subjects)
```

```python
student1 = {"English", "Maths", "CS", "Chemistry", "Physics"}
student2 = {"English", "Biology", "Chemistry", "Physics"}
student3 = {"Sanskrit", "Maths", "CS"}

print(student1, type(student1))
print(student2, type(student2))
print(student3, type(student3))

# all subjects of student1, student2, student3 - union
all_subjects = student1.union(student2, student3)
print(all_subjects)

# or
all_subjects = student1 | student2 | student3
print(all_subjects)
```

```python
student1 = {"English", "Maths", "CS", "Chemistry", "Physics"}
student2 = {"English", "Biology", "Chemistry", "Physics"}
student3 = {"Sanskrit", "Maths", "CS"}

print(student1, type(student1))
print(student2, type(student2))
print(student3, type(student3))

common_subjects = student1.intersection(student2, student3)
print(common_subjects) 
```

```python
# Difference of sets
days = {"Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"}
weekends = {"Saturady", "Sunday"}

weekdays = days - weekends # elements of days which are NOT in weekends
print(weekdays)

# or
weekdays = days.difference(weekends) # elements of days which are NOT in weeends
print(weekdays)
```
