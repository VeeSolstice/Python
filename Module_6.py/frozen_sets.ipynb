```python
# Normal Set - Immutable saets
s1 = {10, 20, 40, 60}
print(s1, type(s1))
s1.add(80)
print(s1, type(s1))
```

```python
# Frozen sets - Immutable sets
fs1 = frozenset({10, 20, 30})
print(fs1, type(fs1))

fs2 = frozenset({10, 50, 100, 200})
print(fs2, type(fs2))

# Operations on Frozenset
print(fs1 & fs2)
print(fs1 | fs2)
print(fs2 - fs1)
```
