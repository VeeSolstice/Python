```python
# Copy -> module
# Shallow copy -> denoted by module.copy -> In this list or dict have diff. memory addresses that they are refering to
                                    #    -> The direct elements NEVER CHANGE(Change made in l1 is NOT reflected to l2)
                                    #    -> The indirect or nested elements CHANGE(Change made in l1 IS reflected to l2)
# Deep copy -> denoted by module.deepcopy -> In this list or dict have diff. memory addresses that they are refering to
                                    #    -> The direct elements NEVER CHANGE(Change made in l1 is NOT reflected to l2)
                                    #    -> The indirect or nested elements NEVER CHANGE(Change made in l1 is NOT reflected to l2)
```

```python
# Shallow Copy in list

import copy
l1 = [1, 2.5, [10,20,30], 'Python']

l2 = copy.copy(l1)

print("Before Change:")
print(f"l1 -> {l1}", id(l1))
print(f"l2 -> {l2}", id(l2))

l1[0] = 5
l1[2][0] =50

print("After Change:")
print(f"l1 -> {l1}", id(l1))
print(f"l2 -> {l2}", id(l2))
```

```python
# Shallow Copy in dictionary

import copy
d1 = {'id': 1111, 'name': 'John', 'marks':{'eng': 71.5, 'maths':91.5, 'bio':80.0}}

d2 = copy.copy(d1)
print("Before Change:")
print(f"l1 -> {d1}", id(d1))
print(f"l2 -> {d2}", id(d2))

d1['name'] = 'Dan'
d1['marks']['maths'] = 92.5

print("After Change:")
print(f"l1 -> {d1}", id(d1))
print(f"l2 -> {d2}", id(d2))
```

```python
# Deep Copy in list

import copy
l1 = [1, 2.5, [10,20,30], 'Python']

l2 = copy.deepcopy(l1)

print("Before Change:")
print(f"l1 -> {l1}", id(l1))
print(f"l2 -> {l2}", id(l2))

l1[0] = 5
l1[2][0] =50

print("After Change:")
print(f"l1 -> {l1}", id(l1))
print(f"l2 -> {l2}", id(l2))
```

```python
# Deep Copy in dictionary

import copy
d1 = {'id': 1111, 'name': 'John', 'marks':{'eng': 71.5, 'maths':91.5, 'bio':80.0}}

d2 = copy.deepcopy(d1)
print("Before Change:")
print(f"l1 -> {d1}", id(d1))
print(f"l2 -> {d2}", id(d2))

d1['name'] = 'Dan'
d1['marks']['maths'] = 92.5

print("After Change:")
print(f"l1 -> {d1}", id(d1))
print(f"l2 -> {d2}", id(d2))
```
