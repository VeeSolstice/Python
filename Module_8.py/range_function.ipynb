-> range() - built-in function used to    generate sequence of integers in a given int

-> range(start, stop, step) stop is not included

-> range(start, stop) => step = 1 by default

-> range()

for i in range(start, stop, step):
    statements

```python
for i in range(1, 11, 1):
     print(i)
```

```python
for i in range(1, 11, 2):
     print(i)
```

```python
# generate even numbes between 1 and 10 (10 excluded)

for i in range(2, 10, 2):
    print(i)
```

```python
# reverse order => 20 to 10 (exculding 10) only even

for i in range(20, 10, -2): # 20, 18, 16, 14, 12
    print(i)
```

```python
# Countdown from 10 to 1 (1 inclluded)

for i in range(10, 0, -1):
    print(i)
print("Happy New Year!!!")    
```

```python
for i in range(5): # start = 0, step = 1 => 0, 1, 2, 3, 4
    print(i)
```

```python
groceries = ['salt', 'milk', 'sugar']

for index in range(len(groceries)): # range(3) => 0, 1, 2
    print(index)
```

```python
profits = [9, 11, 6, 10]

for index in range(len(profits)):
    q = index + 1
    print(f"Profit for quarter {q} is {profits[index]}")
```
