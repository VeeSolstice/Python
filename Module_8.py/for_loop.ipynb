It is an iterator based loop which steps through the items of a collection (lists, tuples, sets, dict, str), and executes a block of code repeatedly for a number of times equal to the items/elements of that collection

for var in sequence:
     statement1
     statement2
     ...
     statementN

```python
percents = [85.5, 81.0, 86.0, 83.5]
print(percents[0])
print(percents[1])
print(percents[2])
print(percents[3])
```

```python
percents = [85.5, 81.0, 86.0, 83.5]

for p in percents:
    print(p)
```
