```python
scores = [2, 45, 102, 4, 9, 12, 45, 90, 1, 0, 1]

total = 0
for score in scores:
    total = total + score
    
print(f"Total runs scored is {total}")
```

```python
scores = [2, 45, 102, 4, 9, 12, 45, 90, 1, 0, 1]

total = sum(scores)
print(f"Total runs scored is {total}")
```

```python
scores = [2, 45, 102, 4, 9, 12, 45, 90, 1, 0, 1]

# Highest score

highest = score

highest = scores[0] #assume that the first value is highest

for score in scores:
    if highest < score:
        highest = score
        
print(f"Highest score is {highest}")
```

```python
scores = [2, 45, 102, 4, 9, 12, 45, 90, 1, 0, 1]

highest = max(scores)
print(f"Highest score is {highest}")
```

```python
scores = [2, 45, 102, 4, 9, 12, 45, 90, 1, 0, 1]

# lowest score

lowest = score

lowest = scores[0] #assume that the first value is lowest

for score in scores:
    if score < lowest:
        lowest = score
        
print(f"Lowest score is {lowest}")
```
