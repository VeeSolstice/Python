import random

randomint(a, b) => returns random int between a and b(both included)

```python
import random
print(random.randint(10, 15))
```

```python
nums =  [10, 4, 1, 8, 4, 3]
# choice(sequence) => returns a random item from the sequence
print(random.choice(nums))
```

```python
fruits = ["Apple", "Orange", "Mango"]

#choice(sequence) => returns a random item from the sequence
print(random.choice(fruits))
```

```python
fruits = ["Apple", "Orange", "Mango"]

#choice(sequence) => returns a random item from the sequence
print(random.shuffle(fruits)) #=> this does not work just returns NONE.
random.shuffle(fruits)
print(fruits)
```
